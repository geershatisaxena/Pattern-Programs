## **Some Pattern Programs in Python**

#### ***Basic Star Patterns***

- Square

```python
n = 10
for i in range(n):
    print("* " * n)
```

- Solid Rectangle

```python
rows = 5
cols = 10
for i in range(rows):
    print("* " * cols)
```

-  Increasing Right Triangle

```python
n = 12
for i in range(1, n + 1):
    print("* " * i)
```

- Decreasing Right Triangle

```python
n = 12
for i in range(n, 0, -1):
    print("* " * i)
```

- Left-Aligned Increasing Triangle

```python
n = 12
for i in range(1, n + 1):
    print("*" * i)
```

- Left-Aligned Decreasing Triangle

```python
n = 12

for i in range(n, 0, -1):
    print("*" * i)
```

- Right-Aligned Triangle

```python
n = 12

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * i)
```

- Inverted Right-Aligned Triangle

```python
n = 12

for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * i)
```

- Pyramid

```python
n = 12
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
```

- Inverted Pyramid

```python
n = 12
for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)
```

#### ***Pyramid Patterns***

- Full Star Pyramid

```python
n = 12
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

- Inverted Full Pyramid

```python
n = 12

for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

- Diamond

```python
n = 12

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))

for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

- Half Diamond

```python
n = 12

for i in range(1, n + 1):
    print("*" * i)

for i in range(n - 1, 0, -1):
    print("*" * i)
```
